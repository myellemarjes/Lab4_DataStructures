```python
from collections import Counter

LAST_NAME = "Marjes"
SEED_NUM = 8
FAVORITE_ARTIST = "The Ridleys"

students = [f"{LAST_NAME}_{i}" for i in range(1, 6)]

subjects = [
    "PROGRAMMING",
    "MATHEMATICS",
    "ELECTRONICS",
    "PHYSICS",
    "ENGLISH"
]

artist_positions = [ord(c) for c in FAVORITE_ARTIST if c.isalpha()]
if not artist_positions:
    artist_positions = [65]

records = []

# Generate student scores
for i, student in enumerate(students):
    for j, subject in enumerate(subjects):
        char_value = artist_positions[(i + j) % len(artist_positions)]
        score = 60 + (
            (SEED_NUM + char_value + i * 7 + j * 11) % 41
        )

        records.append({
            "Student": student,
            "Subject": subject,
            "Score": score
        })

# Add one duplicate record
records.append(records[0].copy())

# Add one incomplete record
records.append({
    "Student": students[0],
    "Subject": "MATHEMATICS",
    "Score": None
})

# Detect duplicate student-subject combinations
seen = set()

for record in records:
    key = (record["Student"], record["Subject"])
    record["Is_Duplicate"] = key in seen
    seen.add(key)

# Categorize performance
def performance_category(score):
    if score is None:
        return "Incomplete"
    elif score >= 90:
        return "Excellent"
    elif score >= 80:
        return "Good"
    elif score >= 75:
        return "Satisfactory"
    else:
        return "Needs Improvement"

for record in records:
    record["Category"] = performance_category(record["Score"])

# Display all records
print("STUDENT PERFORMANCE RECORDS")
for record in records:
    print(record)

# Display duplicates
print("\nDUPLICATE RECORDS")
for record in records:
    if record["Is_Duplicate"]:
        print(record)

# Display incomplete records
print("\nINCOMPLETE RECORDS")
for record in records:
    if record["Score"] is None:
        print(record)

# Summarize student performance
print("\nSTUDENT SUMMARY")

for student in students:
    scores = [
        r["Score"] for r in records
        if r["Student"] == student and r["Score"] is not None
        and not r["Is_Duplicate"]
    ]

    if scores:
        print(
            student,
            "Average:", round(sum(scores) / len(scores), 2),
            "Highest:", max(scores),
            "Lowest:", min(scores)
        )

# Summarize subjects
print("\nSUBJECT SUMMARY")

for subject in subjects:
    scores = [
        r["Score"] for r in records
        if r["Subject"] == subject and r["Score"] is not None
        and not r["Is_Duplicate"]
    ]

    if scores:
        print(
            subject,
            "Average:", round(sum(scores) / len(scores), 2),
            "Highest:", max(scores),
            "Lowest:", min(scores)
        )

# Performance category counts
print("\nPERFORMANCE CATEGORY SUMMARY")

categories = Counter(
    r["Category"] for r in records
    if r["Score"] is not None and not r["Is_Duplicate"]
)

for category, count in categories.items():
    print(category, ":", count)

# Sort records by student and subject
print("\nORDERED RESULTS")

ordered_records = sorted(
    records,
    key=lambda r: (r["Student"], r["Subject"])
)

for record in ordered_records:
    print(record)

# Final report
valid_scores = [
    r["Score"] for r in records
    if r["Score"] is not None and not r["Is_Duplicate"]
]

print("\nFINAL REPORT")
print("Total records:", len(records))
print("Valid scores:", len(valid_scores))
print("Duplicate records:", sum(r["Is_Duplicate"] for r in records))
print("Incomplete records:", sum(r["Score"] is None for r in records))
print("Overall average:", round(sum(valid_scores) / len(valid_scores), 2))
```

```python
LAST_NAME = "Marjes"
SEED_NUM = 8
FAVORITE_ARTIST = "The Ridleys"

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]

prices = {
    "NOTEBOOK": 35.00,
    "PEN": 15.00,
    "USB": 250.00,
    "HEADPHONES": 450.00,
    "CALCULATOR": 180.00
}

products = list(prices.keys())
artist = [ord(c) for c in FAVORITE_ARTIST if c.isalpha()]
if not artist:
    artist = [65]

transactions = []

# Generate transactions
for i, customer in enumerate(customers):
    for j, product in enumerate(products):
        quantity = 1 + (
            SEED_NUM + artist[(i + j) % len(artist)] + i + j
        ) % 5

        transactions.append({
            "Customer": customer,
            "Product": product,
            "Quantity": quantity,
            "Unit Price": prices[product],
            "Total": quantity * prices[product]
        })

# Add duplicate transaction
transactions.append(transactions[0].copy())

# Detect repeated customer-product records
seen = set()
duplicates = []

for transaction in transactions:
    key = (transaction["Customer"], transaction["Product"])
    if key in seen:
        duplicates.append(transaction)
    else:
        seen.add(key)

print("TRANSACTION RECORDS")
for transaction in transactions:
    print(transaction)

print("\nDUPLICATE PURCHASES")
for transaction in duplicates:
    print(transaction)

# Summarize spending per customer
print("\nCUSTOMER SUMMARY")
for customer in customers:
    customer_records = [
        t for t in transactions
        if t["Customer"] == customer
        and (t["Customer"], t["Product"]) not in {
            (d["Customer"], d["Product"]) for d in duplicates
        }
    ]

    total = sum(t["Total"] for t in customer_records)
    quantity = sum(t["Quantity"] for t in customer_records)

    print(customer, "Spending:", total, "Units:", quantity)

# Summarize sales per product
print("\nPRODUCT SUMMARY")
for product in products:
    product_records = [
        t for t in transactions
        if t["Product"] == product
        and (t["Customer"], t["Product"]) not in {
            (d["Customer"], d["Product"]) for d in duplicates
        }
    ]

    quantity = sum(t["Quantity"] for t in product_records)
    sales = sum(t["Total"] for t in product_records)

    print(product, "Quantity sold:", quantity, "Total sales:", sales)

# Final report
valid = [
    t for t in transactions
    if (t["Customer"], t["Product"]) not in {
        (d["Customer"], d["Product"]) for d in duplicates
    }
]

print("\nFINAL TRANSACTION REPORT")
print("Total records:", len(transactions))
print("Duplicate records:", len(duplicates))
print("Distinct customers:", len(set(t["Customer"] for t in transactions)))
print("Distinct products:", len(set(t["Product"] for t in transactions)))
print("Total sales:", sum(t["Total"] for t in valid))
```
